```python
import pandas as pd

# Load the dataset with low_memory=False to suppress DtypeWarning
df = pd.read_csv('/content/accepted_2007_to_2018Q4.csv', low_memory=False)

# Display the first 5 rows of the DataFrame
display(df.head())
```

## 欠損値の確認と処理 (Checking and Handling Missing Values)

データセットの品質を確保するために、まず欠損値の状況を確認し、適切な処理を行います。

```python
# 各列の欠損値の数をカウントし、割合を計算
missing_values = df.isnull().sum()
missing_percentage = (missing_values / len(df)) * 100

# 欠損値がある列のみを抽出し、割合が高い順にソート
missing_info = pd.DataFrame({
    'Missing Count': missing_values,
    'Missing Percentage': missing_percentage
})
missing_info = missing_info[missing_info['Missing Count'] > 0].sort_values(by='Missing Percentage', ascending=False)

print('--- 欠損値のある列と割合 ---')
display(missing_info.head(20)) # 上位20列を表示
```

欠損値の割合が高いカラムが多数存在します。特に`member_id`, `id` はほぼすべて欠損しているため、分析には利用できません。また、`hardship`や`debt_settlement`関連のカラムも多くの欠損が見られます。

これらのカラムは、目的とする「売上や利益を上げるための分析」において直接的な影響が少ない、または欠損が多すぎるため、一旦除外することを検討します。重要なカラムについては、欠損値の補完を検討します。

### ドロップするカラムの選択

欠損値の割合が非常に高い、または分析に直接関係しないと思われる以下のカラムをドロップします。例えば、`member_id`, `id`, `url`, `desc`, `mths_since_last_major_derog`, `hardship_type`, `hardship_reason` などです。これらのカラムは、個人を特定するIDやURL、フリーテキスト、あるいはほとんどのデータで情報がないため、分析に貢献しにくいと判断します。

```python
# 欠損値の割合が90%以上のカラムをリストアップ
columns_to_drop_high_missing = missing_info[missing_info['Missing Percentage'] > 90].index.tolist()

# さらに、直接的な分析に不要と判断されるIDやURLなどのカラムも追加
# (既に90%以上の欠損でリストアップされている可能性もあるが、明示的に指定)
columns_to_drop_manual = [
    'id', 'member_id', 'url', 'desc',
    'hardship_type', 'hardship_reason', 'hardship_status', 'hardship_start_date',
    'hardship_end_date', 'hardship_loan_status', 'hardship_payoff_balance_amount',
    'hardship_last_payment_amount', 'debt_settlement_flag_date', 'settlement_status',
    'settlement_date', 'settlement_amount', 'settlement_percentage', 'settlement_term'
]

# 重複を避けてドロップするカラムを結合
columns_to_drop = list(set(columns_to_drop_high_missing + columns_to_drop_manual))

# DataFrameからカラムをドロップ
df = df.drop(columns=columns_to_drop)

print(f"ドロップしたカラム数: {len(columns_to_drop)}")
print("ドロップ後のDataFrameの形状:", df.shape)

# ドロップ後の欠損値状況を再度確認
missing_values_after_drop = df.isnull().sum()
missing_percentage_after_drop = (missing_values_after_drop / len(df)) * 100
missing_info_after_drop = pd.DataFrame({
    'Missing Count': missing_values_after_drop,
    'Missing Percentage': missing_percentage_after_drop
})
missing_info_after_drop = missing_info_after_drop[missing_info_after_drop['Missing Count'] > 0].sort_values(by='Missing Percentage', ascending=False)

print('\n--- カラムドロップ後の欠損値のある列と割合 ---')
display(missing_info_after_drop.head(10))
```

残りの欠損値は、多くの場合、`mths_since_last_delinq` (最終延滞からの月数) や `mths_since_last_record` (最終記録からの月数) のように、該当する事象がなかったことを示す場合が多いです。これらは「情報がない」という意味で、0や特定の値を埋めることが妥当な場合があります。例えば、延滞経験がない場合は0として扱うなどが考えられます。

ここでは、いくつかの数値型カラムについて、欠損値を0または平均値/中央値で補完します。カテゴリカルなカラムの欠損値は、最も頻繁に出現する値（最頻値）で補完するか、'Unknown' などの新しいカテゴリとして扱います。

### 数値型カラムの補完

*   `mths_since_last_delinq`, `mths_since_recent_revol_delinq`, `mths_since_last_record` など: 0 で補完（該当事象なしと解釈）
*   その他の数値型カラム (`annual_inc`, `revol_util` など): 中央値で補完（外れ値の影響を少なくするため）

### カテゴリカルカラムの補完

*   `emp_title`, `emp_length` など: 'Missing' や 'Unknown' などの新しいカテゴリとして補完

```python
# 数値型カラムの欠損値補完
# 0で補完するカラム
zero_impute_cols = [
    'mths_since_last_delinq', 'mths_since_recent_revol_delinq', 'mths_since_last_record',
    'num_tl_op_past_12m', 'num_rev_accts', 'num_actv_rev_tl', 'num_bc_tl',
    'num_op_rev_tl', 'num_rev_tl_bal_gt_0', 'num_accts_ever_120_pd',
    'num_il_tl', 'num_sats', 'num_bc_sats', 'num_mort_accts', 'num_other_tl',
    'num_undr_rev_tl', 'num_actv_bc_tl', 'num_tl_120dpd_2m', 'num_tl_30dpd',
    'num_tl_90g_dpd_24m', 'pct_tl_nvr_dlq', 'tot_coll_amt', 'tot_cur_bal',
    'avg_cur_bal', 'bc_open_to_buy', 'bc_util', 'mo_sin_old_il_acct',
    'mo_sin_old_rev_tl_op', 'mo_sin_rcnt_rev_tl_op', 'mo_sin_rcnt_tl',
    'mort_acc', 'total_bal_il', 'total_cu_tl', 'total_rev_hi_lim',
    'acc_open_past_24mths', 'chargeoff_within_12_mths', 'delinq_2yrs',
    'inq_last_6mths', 'open_acc', 'pub_rec', 'collections_12_mths_ex_med',
    'tot_hi_cred_lim', 'total_il_high_credit_limit', 'max_bal_bc', 'all_util',
    'inq_fi', 'inq_last_12m', 'open_act_il', 'open_il_12m',
    'open_il_24m', 'open_rv_12m', 'open_rv_24m', 'pct_il_revol',
    'total_bal_ex_mort', 'total_bc_limit', 'total_cc_tl'
]
for col in zero_impute_cols:
    if col in df.columns:
        df[col] = df[col].fillna(0)

# 中央値で補完する数値型カラム（上記以外の残りの数値型欠損値）
for col in df.select_dtypes(include=['float64', 'int64']).columns:
    if df[col].isnull().any():
        median_val = df[col].median()
        df[col] = df[col].fillna(median_val)

# カテゴリカル型カラムの欠損値補完 ('Unknown'で補完)
for col in df.select_dtypes(include='object').columns:
    if df[col].isnull().any():
        df[col] = df[col].fillna('Unknown')

print('\n--- 欠損値補完後の各列の欠損値の数 ---')
display(df.isnull().sum().sort_values(ascending=False).head())

print('\n欠損値処理が完了しました。全ての欠損値が処理されたことを確認してください。')
```

`Fully Paid`になる顧客の特性を理解するために、金利（`int_rate`）がローンの最終ステータスにどのように影響するかを調べます。一般的に、金利が高いローンはリスクが高いとみなされるため、`Charged Off`や`Default`のローンで金利が高くなる傾向があるかを確認します。

## 5. DTI (借入比率) とローンステータス (`loan_status`) の関係 (Relationship between DTI and Loan Status)

DTI (Debt-to-Income Ratio) は、借り手の返済能力を示す重要な指標です。DTIが高い借り手は、既存の債務負担が大きいため、新しいローンを返済する能力が低いとみなされる傾向があります。ここでは、各ローンステータスにおけるDTIの平均値を比較し、DTIがローンのパフォーマンスにどのように影響するかを分析します。

```python
# ローンステータスごとの平均DTIを計算
dti_by_status = df.groupby('loan_status')['dti'].mean().sort_values(ascending=False)

print('--- ローンステータスごとの平均DTI ---')
display(dti_by_status)

plt.figure(figsize=(12, 6))
sns.barplot(x=dti_by_status.index, y=dti_by_status.values, palette='viridis')
plt.title('ローンステータスごとの平均DTI')
plt.xlabel('ローンステータス')
plt.ylabel('平均DTI')
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

## 6. 金利とDTIを組み合わせた顧客セグメント分析 (Combined Analysis of Interest Rate and DTI for Customer Segmentation)

金利とDTIの両方がローンの返済状況に影響を与えることがわかりました。ここでは、これらの2つの指標を組み合わせて顧客をセグメント化し、特に「Fully Paid」になりやすい顧客層を特定します。

```python
# 金利（int_rate）をカテゴリに分割
# 以前の分析結果を参考に、Fully Paidが低い金利帯に集中していることを考慮
# 例: 低金利(<12%), 中金利(12-16%), 高金利(>16%)
bins_int_rate = [0, 12, 16, df['int_rate'].max()]
labels_int_rate = ['Low Interest Rate', 'Medium Interest Rate', 'High Interest Rate']
df['int_rate_segment'] = pd.cut(df['int_rate'], bins=bins_int_rate, labels=labels_int_rate, right=False)

# DTIをカテゴリに分割
# 以前の分析結果を参考に、Fully Paidが低いDTIに集中していることを考慮
# 例: 低DTI(<19%), 中DTI(19-22%), 高DTI(>22%)
bins_dti = [0, 19, 22, df['dti'].max()]
labels_dti = ['Low DTI', 'Medium DTI', 'High DTI']
df['dti_segment'] = pd.cut(df['dti'], bins=bins_dti, labels=labels_dti, right=False)

# セグメントごとのローンステータスの分布を計算
segment_analysis = df.groupby(['int_rate_segment', 'dti_segment', 'loan_status']).size().unstack(fill_value=0)

# 各セグメントにおける「Fully Paid」の割合と「Charged Off」の割合を計算
segment_analysis['Total Loans'] = segment_analysis.sum(axis=1)
segment_analysis['Fully Paid Ratio'] = segment_analysis['Fully Paid'] / segment_analysis['Total Loans']
segment_analysis['Charged Off Ratio'] = segment_analysis['Charged Off'] / segment_analysis['Total Loans']

print('--- 金利とDTIセグメントごとのローンステータス割合 ---')
display(segment_analysis[['Total Loans', 'Fully Paid', 'Fully Paid Ratio', 'Charged Off', 'Charged Off Ratio']].sort_values(by='Fully Paid Ratio', ascending=False))
```

この表から、「Low Interest Rate」かつ「Low DTI」のセグメントが最も「Fully Paid Ratio」が高く、リスクが低い顧客層であることがわかります。一方で、「High Interest Rate」かつ「High DTI」のセグメントでは「Charged Off Ratio」が高くなる傾向が見られます。

これらのセグメントをヒートマップで視覚化し、より直感的に理解を深めましょう。

```python
import numpy as np

# Fully Paid Ratioのヒートマップ
plt.figure(figsize=(12, 6))
fully_paid_heatmap_data = segment_analysis['Fully Paid Ratio'].unstack()
sns.heatmap(fully_paid_heatmap_data, annot=True, cmap='Greens', fmt='.2%', linewidths=.5, linecolor='black')
plt.title('金利とDTIセグメントごとのFully Paid率')
plt.xlabel('DTIセグメント')
plt.ylabel('金利セグメント')
plt.xticks(rotation=45)
plt.yticks(rotation=0)
plt.tight_layout()
plt.show()

# Charged Off Ratioのヒートマップ
plt.figure(figsize=(12, 6))
charged_off_heatmap_data = segment_analysis['Charged Off Ratio'].unstack()
sns.heatmap(charged_off_heatmap_data, annot=True, cmap='Reds', fmt='.2%', linewidths=.5, linecolor='black')
plt.title('金利とDTIセグメントごとのCharged Off率')
plt.xlabel('DTIセグメント')
plt.ylabel('金利セグメント')
plt.xticks(rotation=45)
plt.yticks(rotation=0)
plt.tight_layout()
plt.show()
```

## 8. 勤務年数 (`emp_length`) とFully Paid率の相関 (Correlation between Employment Length and Fully Paid Ratio)

勤務年数は、顧客の雇用安定性を示す重要な指標であり、ローンの返済能力と関連している可能性があります。ここでは、勤務年数ごとのFully Paid率を分析し、安定した雇用がローンの完済にどのように影響するかを確認します。

```python
# 'emp_length'のカテゴリ順序を定義 (Unknownは最後)
emp_length_order = [
    '< 1 year', '1 year', '2 years', '3 years', '4 years', '5 years',
    '6 years', '7 years', '8 years', '9 years', '10+ years', 'Unknown'
]

# emp_lengthとloan_statusでグループ化し、各ステータスの件数をカウント
emp_length_status_counts = df.groupby(['emp_length', 'loan_status']).size().unstack(fill_value=0)

# 各emp_lengthにおける総ローン数とFully Paid数を計算
emp_length_status_counts['Total Loans'] = emp_length_status_counts.sum(axis=1)
emp_length_status_counts['Fully Paid Ratio'] = emp_length_status_counts['Fully Paid'] / emp_length_status_counts['Total Loans']

# Fully Paid Ratioが高い順にソート（Unknownは除く）
fully_paid_ratio_by_emp_length = emp_length_status_counts[['Total Loans', 'Fully Paid', 'Fully Paid Ratio']].sort_values(by='Fully Paid Ratio', ascending=False)

print('--- 勤務年数ごとのFully Paid率 ---')
display(fully_paid_ratio_by_emp_length)

# 勤務年数の順序でプロットするために、データフレームを並べ替える
plot_data = emp_length_status_counts.loc[emp_length_order].dropna(subset=['Fully Paid Ratio']) # Unknownを除くか、最後に表示する

plt.figure(figsize=(12, 6))
sns.barplot(x=plot_data.index, y='Fully Paid Ratio', data=plot_data, palette='viridis')
plt.title('勤務年数ごとのFully Paid率')
plt.xlabel('勤務年数')
plt.ylabel('Fully Paid 率')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## 9. 借入目的 (`purpose`) とFully Paid率の相関 (Correlation between Loan Purpose and Fully Paid Ratio)

借入目的は、借り手の行動や経済状況を反映する重要な要素です。ここでは、各借入目的ごとのFully Paid率を分析し、どのような目的のローンが完済されやすいか、あるいはリスクが高いかを評価します。

```python
# purposeとloan_statusでグループ化し、各ステータスの件数をカウント
purpose_status_counts = df.groupby(['purpose', 'loan_status']).size().unstack(fill_value=0)

# 各purposeにおける総ローン数とFully Paid数を計算
purpose_status_counts['Total Loans'] = purpose_status_counts.sum(axis=1)
purpose_status_counts['Fully Paid Ratio'] = purpose_status_counts['Fully Paid'] / purpose_status_counts['Total Loans']
purpose_status_counts['Charged Off Ratio'] = purpose_status_counts['Charged Off'] / purpose_status_counts['Total Loans']

# Fully Paid Ratioが高い順にソート
fully_paid_ratio_by_purpose = purpose_status_counts[['Total Loans', 'Fully Paid', 'Fully Paid Ratio', 'Charged Off', 'Charged Off Ratio']].sort_values(by='Fully Paid Ratio', ascending=False)

print('--- 借入目的ごとのFully Paid率とCharged Off率 ---')
display(fully_paid_ratio_by_purpose)

plt.figure(figsize=(15, 7))
sns.barplot(x=fully_paid_ratio_by_purpose.index, y='Fully Paid Ratio', data=fully_paid_ratio_by_purpose, palette='viridis')
plt.title('借入目的ごとのFully Paid率')
plt.xlabel('借入目的')
plt.ylabel('Fully Paid 率')
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

## 10. リスクの高い借入目的 (`purpose`) に特化した分析 (Specialized Analysis for High-Risk Loan Purposes)

先ほどの分析で、`small_business` と `house` が他の借入目的に比べて`Charged Off`率が高いことがわかりました。ここでは、これらのリスクの高い目的に絞り込み、金利（`int_rate`）やDTI（`dti`）といった主要なリスク指標が、`Fully Paid`と`Charged Off`のローンでどのように異なるかを詳細に分析します。これにより、これらの高リスクカテゴリにおける審査基準の改善に役立つ洞察を得ます。

```python
# リスクの高い目的のローンを抽出
high_risk_purposes = ['small_business', 'house']
df_high_risk = df[df['purpose'].isin(high_risk_purposes)]

print('--- リスクの高い借入目的（small_business, house）のローン統計 ---')
display(df_high_risk.groupby(['purpose', 'loan_status'])[['int_rate', 'dti', 'fico_range_high']].mean().unstack())

# 金利とDTIの分布を可視化（Fully Paid vs Charged Off）
plt.figure(figsize=(18, 12))

# small_businessの金利分布
plt.subplot(2, 2, 1)
sns.histplot(data=df_high_risk[df_high_risk['purpose'] == 'small_business'], x='int_rate', hue='loan_status', kde=True, palette='coolwarm')
plt.title('Small Businessローンの金利分布 (Fully Paid vs Charged Off)')
plt.xlabel('金利 (%)')
plt.ylabel('件数')

# small_businessのDTI分布
plt.subplot(2, 2, 2)
sns.histplot(data=df_high_risk[df_high_risk['purpose'] == 'small_business'], x='dti', hue='loan_status', kde=True, palette='coolwarm')
plt.title('Small BusinessローンのDTI分布 (Fully Paid vs Charged Off)')
plt.xlabel('DTI')
plt.ylabel('件数')

# houseの金利分布
plt.subplot(2, 2, 3)
sns.histplot(data=df_high_risk[df_high_risk['purpose'] == 'house'], x='int_rate', hue='loan_status', kde=True, palette='coolwarm')
plt.title('Houseローンの金利分布 (Fully Paid vs Charged Off)')
plt.xlabel('金利 (%)')
plt.ylabel('件数')

# houseのDTI分布
plt.subplot(2, 2, 4)
sns.histplot(data=df_high_risk[df_high_risk['purpose'] == 'house'], x='dti', hue='loan_status', kde=True, palette='coolwarm')
plt.title('HouseローンのDTI分布 (Fully Paid vs Charged Off)')
plt.xlabel('DTI')
plt.ylabel('件数')

plt.tight_layout()
plt.show()
```

## 11. 年収 (`annual_inc`) とローンステータス (`loan_status`) の関係 (Relationship between Annual Income and Loan Status)

## 融資ポートフォリオ再構築案レポート：収益性最大化のための戦略

### 1. はじめに

本レポートは、ローンデータ分析の結果に基づき、当社の融資ポートフォリオの収益性を最大化するための具体的な再構築案を提示します。これまでの分析により、特定の顧客セグメントやローンの特性が「Fully Paid」（完済）率や「Charged Off」（貸倒）率に大きな影響を与えることが明らかになりました。

### 2. 主要な分析結果の要約

*   **金利（int_rate）、DTI（dti）、FICOスコア（fico_range_high）の重要性**: 「Fully Paid」となる顧客は、平均的に低金利、低DTI、高FICOスコアの傾向がありました。特に「Low Interest Rate」かつ「Low DTI」のセグメントは「Fully Paid」率が85.66%と非常に高く、最も優良な顧客層であることが判明しました。一方、「High Interest Rate」かつ「High DTI」のセグメントでは「Charged Off」率が高い傾向が見られました。
*   **借入目的（purpose）のリスク**: `small_business`と`house`目的のローンは、他の目的と比較して「Charged Off」率が有意に高く、リスクが高いカテゴリとして特定されました。これらの目的のローンでは、平均金利が高く、平均DTIも高い傾向がありました。
*   **勤務年数（emp_length）と年収（annual_inc）**: 勤務年数は金利やDTIほど明確な相関は見られませんでしたが、「Unknown」は最も「Fully Paid」率が低い結果でした。年収に関しては、高年収でも「Default」が発生するケースがあり、低年収層でも完済するケースがあるなど、単独での判断は難しいことが示唆されました。

### 3. 収益性最大化のための課題と問題点

1.  **高リスクセグメントへの不適切な融資**: 高金利・高DTIの顧客層や、`small_business`・`house`といった高リスクの借入目的への融資において、十分なリスク評価が行われていない可能性があります。これにより、「Charged Off」率が高まり、回収不能な損失が発生しています。
2.  **優良顧客セグメントの特定と最大化の不足**: 低金利・低DTIの優良顧客層を積極的に特定し、融資を拡大する戦略が十分に実施されていない可能性があります。
3.  **リスク評価基準の不均一性**: 金利、DTI、FICOスコアといった重要な指標が個別に評価され、それらを統合的に判断するリスクスコアリングモデルの運用が最適化されていない可能性があります。

### 4. 融資ポートフォリオ再構築案

上記の課題を解決し、収益性を最大化するために、以下のポートフォリオ再構築案を提案します。

1.  **優良顧客セグメントへの融資集中と拡大**:
    *   **「Low Interest Rate (<12%)」かつ「Low DTI (<19%)」の顧客層**への融資を最優先し、マーケティングリソースを集中させます。これらの顧客は「Fully Paid」率が最も高く、安定した利息収入を期待できます。
    *   **信用力の高い顧客（FICOスコア700以上）**に対しては、競争力のある金利を提供し、他社からの顧客獲得を強化します。

2.  **高リスク借入目的への審査厳格化と差別化**:
    *   **`small_business`および`house`目的のローン**に対しては、**金利が18%を超える融資**、または**DTIが18%を超える融資**を特に厳格に審査します。
    *   これらの目的のローンでは、**FICOスコアの最低基準を700以上**に引き上げ、さらに詳細な財務状況の審査を追加します。
    *   リスクが高いと判断されるケースでは、融資額の制限、追加担保の要求、あるいは融資自体の見送りを検討します。

3.  **総合的なリスクスコアリングモデルの導入と最適化**:
    *   金利、DTI、FICOスコア、年収、勤務年数、借入目的など、複数の要因を総合的に評価する**機械学習ベースのリスクスコアリングモデル**を導入します。これにより、個々の顧客に対するリスクをより正確に予測し、融資判断の精度を高めます。
    *   モデルは定期的に再学習・更新し、市場環境や顧客動向の変化に対応できるようにします。

4.  **金利設定の最適化**:
    *   リスクプロファイルに応じて金利を動的に設定する**ダイナミックプライシング戦略**を導入します。これにより、優良顧客には低金利で魅力的なオファーを、高リスク顧客にはリスクに見合った適切な金利を設定し、収益とリスクのバランスを最適化します。

5.  **モニタリングと早期警戒システムの強化**:
    *   ローン実行後も顧客の返済状況や経済状況を継続的にモニタリングし、返済遅延の兆候が見られた場合には、早期に介入できるシステム（早期警戒システム）を強化します。これにより、「Late」や「Charged Off」への移行を最小限に抑えます。

### 5. 期待される効果

本再構築案を実施することで、以下の効果が期待されます。

*   **「Fully Paid」率の向上と「Charged Off」率の低減**: より健全な融資判断により、回収不能な損失が減少し、安定した利息収入が増加します。
*   **収益性の向上**: 優良顧客への融資拡大と高リスク顧客への適切な対応により、融資ポートフォリオ全体の利益率が向上します。
*   **リスク管理の強化**: 総合的なリスク評価と早期警戒システムにより、将来的な金融リスクへの対応力が強化されます。
